```python
from dotenv import load_dotenv
import os
import requests

load_dotenv()
API_KEY_AEMET = os.getenv("API_KEY_AEMET")

# La api_key va como query string, no en la cabecera.
res = requests.get(
    url = 'https://opendata.aemet.es/opendata/api/valores/climatologicos/inventarioestaciones/todasestaciones',
    params = {"api_key": API_KEY_AEMET},
    timeout = 30,
)
sobre = res.json()
print(sobre)  # sobre de la API: estado, descripcion y enlace "datos"

# Seguir el enlace "datos" para descargar el inventario real.
res_datos = requests.get(sobre["datos"], timeout = 30)
res_datos.encoding = "ISO-8859-15"
estaciones = res_datos.json()
print(next(e for e in estaciones if e["nombre"] == "MADRID, CIUDAD UNIVERSITARIA"))
```
