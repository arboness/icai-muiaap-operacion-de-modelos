```python
from dotenv import load_dotenv
import os
import requests

load_dotenv()
API_KEY_AEMET = os.getenv("API_KEY_AEMET")



res = requests.get(
    url = 'https://opendata.aemet.es/opendata/api/valores/climatologicos/inventarioestaciones/todasestaciones',
    headers = {
        "api_key": API_KEY_AEMET
    }
)

print(res.json())

```
